```python
import os
import torch
import gc
from transformers import LlavaNextProcessor, LlavaNextForConditionalGeneration

# Set cache directory (shared with other models)
CACHE_DIR = os.path.abspath("cache_dir")
os.environ['HUGGINGFACE_HUB_CACHE'] = CACHE_DIR

# Enable efficient GPU memory management
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "max_split_size_mb:512"
torch.backends.cuda.matmul.allow_tf32 = True

# Device setup
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch_dtype = torch.bfloat16 if torch.cuda.is_available() else torch.float32

# Model ID and model name (for output naming)
model_id = "llava-hf/llava-v1.6-vicuna-7b-hf"
model_name = "LLaVA"

# Load processor and model if not yet loaded
if "processor" not in globals():
    print("Loading LLaVA processor...")
    processor = LlavaNextProcessor.from_pretrained(model_id, cache_dir=CACHE_DIR)

if "model" not in globals():
    print("Loading LLaVA model...")
    model = LlavaNextForConditionalGeneration.from_pretrained(
        model_id,
        torch_dtype=torch_dtype,
        low_cpu_mem_usage=True,
        cache_dir=CACHE_DIR
    ).to(device, non_blocking=True)
    print("LLaVA is ready!")

# Clean memory
torch.cuda.empty_cache()
gc.collect()
```

```python
import csv
from pynvml import (
    nvmlInit, nvmlDeviceGetHandleByIndex, nvmlDeviceGetUtilizationRates,
    nvmlDeviceGetMemoryInfo, nvmlDeviceGetPowerUsage, nvmlShutdown
)

# Initialize GPU tracking (once)
nvmlInit()
gpu_handle = nvmlDeviceGetHandleByIndex(0)

def log_gpu_usage():
    util = nvmlDeviceGetUtilizationRates(gpu_handle)
    mem = nvmlDeviceGetMemoryInfo(gpu_handle)
    power = nvmlDeviceGetPowerUsage(gpu_handle)
    return util.gpu, mem.used / 1024**2, power / 1000  # %, MB, W
```

```python
import time
import cv2
from PIL import Image


def process_video(video_path, results_dir, prompt, model, processor, device):


    video_file = os.path.basename(video_path)
    video_name = os.path.splitext(video_file)[0]

    frames_dir = os.path.join(results_dir, "frames")
    os.makedirs(results_dir, exist_ok=True)
    os.makedirs(frames_dir, exist_ok=True)

    log_file_path = os.path.join(results_dir, "results.txt")
    csv_log_path = os.path.join(results_dir, "performance_log.csv")
    csv_header_written = os.path.isfile(csv_log_path) and os.path.getsize(csv_log_path) > 0

    cap = cv2.VideoCapture(video_path, cv2.CAP_FFMPEG)
    if not cap.isOpened():
        print(f"Failed to open video: {video_file}")
        return

    fps = cap.get(cv2.CAP_PROP_FPS)
    interval = int(1 * fps)
    print(f"Processing '{video_file}' with prompt: [{prompt}]")

    descriptions = []
    frame_counter = 0
    actual_frame_counter = 0
    yes_frames = []

    try:
        while cap.isOpened():
            success, frame = cap.read()
            if not success:
                break

            actual_frame_counter += 1
            frame_time = actual_frame_counter / fps

            if actual_frame_counter % interval == 0:
                frame_counter += 1
                start_time = time.time()

                frame_pil = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))

                conversation = [
                    {
                        "role": "user",
                        "content": [
                            {"type": "text", "text": prompt},
                            {"type": "image"},
                        ],
                    }
                ]
                raw_question = conversation[0]["content"][0]["text"]
                token_count = len(raw_question.split())
                chat = processor.apply_chat_template(conversation, add_generation_prompt=True)

                inputs = processor(images=frame_pil, text=chat, return_tensors="pt").to(device)

                with torch.no_grad():
                    outputs = model.generate(
                        **inputs,
                        return_dict_in_generate=True,
                        output_scores=True,
                        max_new_tokens=2,
                        do_sample=False,
                        num_beams=2,
                        repetition_penalty=1.5,
                    )

                response = processor.batch_decode(outputs.sequences, skip_special_tokens=True)[0]
                response = response.replace(raw_question, "").replace("USER:", "").replace("ASSISTANT:", "").strip()
                response = "Yes" if "Yes" in response else "No" if "No" in response else "Unclear"

                logits = outputs.scores
                confidence = sum(torch.nn.functional.softmax(logit, dim=-1).max().item() for logit in logits) / len(logits)

                if response == "Yes":
                    yes_frames.append((confidence, actual_frame_counter, frame_pil.copy()))

                inference_time = (time.time() - start_time) * 1000

                gpu_usage, vram_mb, power_w = log_gpu_usage()

                with open(csv_log_path, "a", newline="") as f:
                    writer = csv.writer(f)
                    if not csv_header_written:
                        writer.writerow([
                            "Frame", "Time (s)", "Prompt", "Model", "Tokens",
                            "GPU Usage (%)", "VRAM (MB)", "Power (W)",
                            "Inference Time (ms)", "Confidence"
                        ])
                        csv_header_written = True

                    writer.writerow([
                        actual_frame_counter, f"{frame_time:.2f}", prompt, "LLaVA", token_count,
                        f"{gpu_usage:.2f}", f"{vram_mb:.2f}", f"{power_w:.2f}",
                        f"{inference_time:.2f}", f"{confidence:.2f}"
                    ])

                result_line = f"Frame {actual_frame_counter} (Time: {frame_time:.2f}s) - Confidence: {confidence:.2f} - {response} - Processing Time: {inference_time:.2f}ms"
                descriptions.append(result_line)
                with open(log_file_path, "a") as log_file:
                    log_file.write(result_line + "\n")

                print(f"  Frame {frame_counter} done - {response} - {confidence:.2f} conf")

                del inputs, outputs
                torch.cuda.empty_cache()
                gc.collect()

        if yes_frames:
            yes_frames.sort(key=lambda x: x[0], reverse=True)
            top_conf = []
            for i, (conf, idx, pil_img) in enumerate(yes_frames):
                if i < 10:
                    top_conf.append((conf, idx, pil_img))
                elif conf == top_conf[-1][0]:
                    top_conf.append((conf, idx, pil_img))
                else:
                    break

            for conf, idx, pil_img in top_conf:
                filename = os.path.join(frames_dir, f"frame_{idx:05d}_conf_{conf:.2f}.jpg")
                pil_img.save(filename)

            print(f"  Saved {len(top_conf)} top 'Yes' frames.")
        else:
            print("  No 'Yes' frames found.")

    except Exception as e:
        print(f"Error in video '{video_file}': {e}")

    cap.release()
```

```python
# Base dataset and results folder
video_root_folder = "/home/saverio/Desktop/surv/Videos"
base_results_dir = "results_llava_tot"
os.makedirs(base_results_dir, exist_ok=True)

difficulty_levels = ["easy", "medium", "hard"]
prompt_templates = {
    "easy":   "Do you see an act of {category} in the frame? Answer Yes or No.",
    "medium": "Does the frame show something that could be interpreted as an instance of {category}? Answer Yes or No.",
    "hard":   "Could a human observer reasonably conclude that this frame depicts {category}-like behavior or consequences? Answer Yes or No."
}

# Loop through each class/category folder
for name_class in sorted(os.listdir(video_root_folder)):
    class_path = os.path.join(video_root_folder, name_class)
    if not os.path.isdir(class_path):
        continue

    print(f"\n--- Processing category: {name_class} ---")

    for difficulty in difficulty_levels:
        prompt = prompt_templates[difficulty].format(category=name_class.lower())

        video_files = sorted([
            f for f in os.listdir(class_path)
            if f.lower().endswith(".mp4") or f.lower().endswith(".avi")
        ])

        if not video_files:
            print(f"  No videos found in: {class_path}")
            continue

        print(f"  Found {len(video_files)} video(s) for difficulty: {difficulty}")

        for video_file in video_files:
            video_path = os.path.join(class_path, video_file)
            video_name = os.path.splitext(video_file)[0]

            results_dir = os.path.join(base_results_dir, name_class, difficulty, video_name)
            process_video(video_path, results_dir, prompt, model, processor, device)

# Shutdown GPU tracking
nvmlShutdown()
```
